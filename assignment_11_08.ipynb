```python
#Given:
#salary = np.array([25000, 32000, 28000, 45000,
 #                  50000, 38000, 42000, 30000])
#Find:

#Total salary
#Average salary
#Highest salary
#Lowest salary
#Median salary
#Standard deviation
salary = np.array([25000, 32000, 28000, 45000,
                  50000, 38000, 42000, 30000])
print("Total salary:", np.sum(salary))
print("Average salary:", np.mean(salary))
print("Highest salary:", np.max(salary))
print("Lowest salary:", np.min(salary))
print("Median salary:", np.median(salary))
print("Standard deviation:", np.std(salary))
```

```python
#Given: marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
#Using NumPy boolean indexing, find:
#Marks greater than 75
#Marks between 50 and 80
#Number of students who scored below 50
#Percentage of students who scored 60 or above
marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
print("Marks > 75:", marks[marks > 75])
print("Marks btwn 50 and 80:", marks[(marks >= 50) & (marks <= 80)])
print("No.of students scored below 50:", np.sum(marks < 50))
print("Percentage of students scored 60 or greater:", np.mean(marks >= 60) * 100)
```

```python
#The following array contains marks of four students in three subjects:
#marks = np.array([
 #   [80, 75, 90],
  #  [65, 70, 72],
   # [88, 92, 85],
    #[55, 60, 58]])
#Calculate:
#Total marks of each student
#Average marks of each student
#Average marks in each subject
#Student with the highest total
#Highest mark in each subject.
marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])
print("Total marks of each student:", np.sum(marks, axis=1))
print("Average marks of each student:", np.mean(marks, axis=1))
print("Average marks in each subject:", np.mean(marks, axis=0))
print("Student with the highest total:", np.argmax(np.sum(marks, axis=1)))
print("Highest mark in each subject:", np.max(marks, axis=0))
```

```python
#11. Create:
#A = np.array([
 #   [1, 2],
  #  [3, 4]])
#B = np.array([
 #   [5, 6],
  #  [7, 8]])
#Perform:
#Matrix addition
#Element-wise multiplication
#Matrix multiplication
#Transpose of A.
#explain the difference between:
#A * B and A @ B
A = np.array([
    [1, 2],
    [3, 4]])
B = np.array([
    [5, 6],
    [7, 8]])
print("Matrix addition:\n", A + B)
print("Element-wise multiplication:\n", A * B)  
print("Matrix multiplication:\n", A @ B)
print("Transpose of A:\n", A.T)
print("A * B is element-wise multiplication, while A @ B is matrix multiplication.")
```
