```python
#Given: marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
#Using NumPy boolean indexing, find:
#Marks greater than 75
#Marks between 50 and 80
#Number of students who scored below 50
#Percentage of students who scored 60 or above
marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
print("Marks > 75:", marks[marks > 75])
print("Marks btwn 50 and 80:", marks[(marks >= 50) & (marks <= 80)])
print("No.of students scored below 50:", np.sum(marks < 50))
print("Percentage of students scored 60 or greater:", np.mean(marks >= 60) * 100)
```

```python
#12. Daily temperatures for two weeks are:
#temperature = np.array([
 #   31, 32, 33, 35, 36, 34, 32,
  #  30, 29, 31, 33, 35, 37, 36])
#Find:
#Mean temperature
#Maximum and minimum temperatures
#Number of days above the average temperature
#Temperatures greater than 34°C
#Difference between maximum and minimum
#Standard deviation
import numpy as np
temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])
print("Mean temperature:", np.mean(temperature))
print("Max Temp:", np.max(temperature))
print("Min Temp:", np.min(temperature))
print("No.of days above avg:", np.sum(temperature > np.mean(temperature)))
print("Temperatures greater than 34°C:", temperature[temperature > 34])
print("Difference between max and min:", np.max(temperature) - np.min(temperature))
print("Standard deviation:", np.std(temperature))
```

```python
"""Consider marks of five students in four subjects:
marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]])
Write a NumPy program that calculates:
Total marks of every student
Average of every student
Average mark for every subject
Highest-scoring student
Lowest-scoring student
Students whose average is greater than 80
Overall class average
Then assign a performance category based on average:
|  Average | Category          |
| -------: | ----------------- |
|     ≥ 85 | Excellent         |
| 70–84.99 | Good              |
| 50–69.99 | Average           |
|     < 50 | Needs Improvement |

Challenge: Try assigning these categories using np.where() or np.select() instead of a Python loop.
"""
marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]])
print("Total marks of every student:", np.sum(marks, axis=1))
print("Average of every student:", np.mean(marks, axis=1))
print("Average mark for every subject:", np.mean(marks, axis=0))
print("Highest-scoring student:", np.argmax(np.sum(marks, axis=1)) + 1)
print("Lowest-scoring student:", np.argmin(np.sum(marks, axis=1)) + 1)
print("Students whose average is greater than 80:", np.sum(np.mean(marks, axis=1) > 80))
print("Overall class average:", np.mean(np.mean(marks, axis=1)))
print("Performance categories:")
average_scores = np.mean(marks, axis=1)
categories = np.select(
    [average_scores >= 85, average_scores >= 70, average_scores < 50],
    ['Excellent', 'Good', 'Needs Improvement'],
    default='Average'
)
print(categories)
```

```python
#21. Take your own example to normalize an array.
array = np.array([1, 2, 3, 4, 5])
normalized_array = (array - np.min(array)) / (np.max(array) - np.min(array))
print("Original array:", array)
print("Normalized array:", normalized_array)
```

```python
#22. use your own data to standardize an array. 
array = np.array([1, 2, 3, 4, 5])
standardized_array = (array - np.mean(array)) / np.std(array)
print("Original array:", array) 
print("Standardized array:", standardized_array)
```
