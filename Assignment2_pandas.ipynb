```python
# 1. Create a Pandas Series containing the marks: 78, 85, 92, 67, 88

#     Display:

# 	Complete Series
# 	Values
# 	Index
# 	Data type
# 	Mean
# 	Maximum
# 	Minimum
marks = pd.Series([78, 85, 92, 67, 88])
print("Marks:", marks)
print("Values:", marks.values)
print("Index:", marks.index)
print("Data Type:", marks.dtype)
print("Mean:", marks.mean())
print("Maximum:", marks.max())
print("Minimum:", marks.min())
```

```python
# 2. Create a Series containing:
 
#    	| Student | Marks |
# 	| ------- | ----: |
# 	| Asha    |    85 |
# 	| Ravi    |    78 |
# 	| Meena   |    92 |
# 	| Kiran   |    74 |
# 	| Arun    |    88 |
# 	Display the marks of Meena.
student_marks = pd.Series([85, 78, 92, 74, 88], index=['Asha', 'Ravi', 'Meena', 'Kiran', 'Arun'])
print("Marks of Meena:", student_marks['Meena'])
```

```python
# 3. Create the following DataFrame:

# | Name  | Age | Department | Marks |
# | ----- | --: | ---------- | ----: |
# | Asha  |  22 | CSE        |    85 |
# | Ravi  |  23 | ECE        |    78 |
# | Meena |  21 | CSE        |    92 |
# | Kiran |  24 | ME         |    74 |
# | Arun  |  22 | ECE        |    88 |

# Display:

# 	Complete DataFrame
# 	First 3 rows
# 	Last 2 rows
# 	Shape
# 	Column names
# 	Data types
df=pd.DataFrame({
    'Name': ['Asha', 'Ravi', 'Meena', 'Kiran', 'Arun'],
    'Age': [22, 23, 21, 24, 22],
    'Department': ['CSE', 'ECE', 'CSE', 'ME', 'ECE'],
    'Marks': [85, 78, 92, 74, 88]
})
print("Complete DataFrame:\n", df)
print("\nFirst 3 rows:\n", df.head(3))
print("\nLast 2 rows:\n", df.tail(2))
print("\nShape:", df.shape)
print("\nColumn names:", df.columns)
print("\nData types:\n", df.dtypes)
```

```python
# 4. Using the previous DataFrame, display:

# 	info()
# 	describe()
# 	Number of rows
# 	Number of columns
print("\nInfo:\n",df.info())
print("\nDescribe:\n",df.describe())
print("\nNo.of rows:", df.shape[0])
print("\nNo.of columns:", df.shape[1])
```

```python
# 5. Using df, display:

# 	Name column
# 	Marks column
# 	Name and Marks together
print("\nName column:\n", df['Name'])
print("\nMarks column:\n", df['Marks'])
print("\nName and Marks together:\n", df[['Name', 'Marks']])
```

```python
    # 6. Display:

    # 	First row using iloc
    # 	Third row using iloc
    # 	Rows 1 to 3
    # 	Name and Marks columns using loc
print("\n1st row using iloc:\n", df.iloc[0])
print("\n3rd row using iloc:\n", df.iloc[2])
print("\nRows 1 to 3:\n", df.iloc[1:4])
print("\nName and Marks columns using loc:\n", df.loc[:, ['Name', 'Marks']])
```

```python
# 7. Find:

# 	Students scoring above 80.
# 	Students scoring between 75 and 90.
# 	Students belonging to CSE.
# 	CSE students scoring above 80.
print("Students scoring >80:\n",df[df['Marks']>80])
print("Students scoring between 75 and 90:\n",df[(df['Marks']>75) & (df['Marks']<90)])
print("Students belonging to CSE:\n",df[df['Department']=='CSE'])
print("CSE students scoring above 80:\n",df[(df['Department']=='CSE') & (df['Marks']>80)])
```

```python
# 8. Display students belonging to either CSE or ECE.
print("Students belonging to either CSE or ECE:\n", df[(df['Department']=='CSE') | (df['Department']=='ECE')])
```

```python
# # 9. Add a new column called Result. Students scoring 75 or above should receive "Pass", otherwise "Fail".
df['Result']=df['Marks'].apply(lambda x: 'Pass' if x>=75 else 'Fail')
df
```

```python
# 10. Assign grades according to:

# 	| Marks | Grade |
# 	| ----: | ----- |
# 	|  ≥ 90 | A     |
# 	|  ≥ 80 | B     |
# 	|  ≥ 70 | C     |
# 	|  < 70 | D     |
df['Grade'] = pd.cut(df['Marks'], bins=[0, 70, 80, 90, 100], labels=['D', 'C', 'B', 'A'], right=False)
df
```

```python
# 11. Sort students:

# 	By marks in ascending order.
# 	By marks in descending order.
# 	By Department and then Marks.
df1=df.sort_values(by='Marks',ascending=True)
df2=df.sort_values(by='Marks',ascending=False)
df3=df.sort_values(by=['Department', 'Marks'],ascending=[True, True])
print("\nSorted by marks in ascending order:\n", df1)
print("\nSorted by marks in descending order:\n", df2)
print("\nSorted by Department and then Marks:\n", df3)
```

```python
# 12. Consider:

# data = {
#     "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
#     "Age": [22, 23, np.nan, 24, 22],
#     "Marks": [85, np.nan, 92, 74, 88]
# }
# students = pd.DataFrame(data)
# Find:
# Missing values
# Number of missing values in each column
# Rows containing missing values
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}
students = pd.DataFrame(data)
print("\nMissing values:\n", students.isnull())
print("\nNumber of missing values in each column:\n", students.isnull().sum())
print("\nRows containing missing values:\n", students[students.isnull().any(axis=1)])
```

```python
# 13. Replace:

# 	Missing Age with average Age
# 	Missing Marks with average Marks
students['Age'] = students['Age'].fillna(students['Age'].mean())
students['Marks'] = students['Marks'].fillna(students['Marks'].mean())
students
```
