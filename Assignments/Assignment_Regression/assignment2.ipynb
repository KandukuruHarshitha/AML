```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, r2_score,mean_absolute_error
import numpy as np
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
df=pd.read_csv(r"C:\Users\bda\Desktop\AML\student_data.csv")
df
```

```python
x=df[["Study Hours_day", "Attendeance", "Assignment_Score"]]
y=df["Final_Marks"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
```

```python
print("Intercept:", model.intercept_)
print("Slope:", model.coef_[0])
print("y_actual Values:", y_test.values)
print("y_predicted Values:", y_pred)
```

```python
print("MSE:", mean_squared_error(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R-squared:", r2_score(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt
 
plt.scatter(y_test, y_pred, color='blue')
minimum=min(y_test.min(), y_pred.min())
maximum=max(y_test.max(), y_pred.max())
plt.plot([minimum, maximum], [minimum, maximum], color='red', linestyle='--')
plt.xlabel('Actual Marks')
plt.ylabel('Predicted Marks')
plt.title('Actual vs Predicted Marks')
plt.show()
```

```python
residuals = y_test - y_pred
plt.scatter(y_pred, residuals, color='green')
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel('Predicted Marks')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.grid(True)
plt.show()
```
