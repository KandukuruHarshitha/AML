```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data={
    "Study Hours_day":[2, 3, 4, 5,6,7,8,9],
    "Attendeance":[60,65,70,75,80,85,90,95],
    "Assignment_Score":[50,55,58,65,70,74,80,88],
    "Final_Marks":[42,48,54,61,67,73,81,89]
}
```

```python
df=pd.DataFrame(data)
df['Study_Hours_week'] = df['Study Hours_day'] * 7
print(df)
```

```python
x=df[["Study Hours_day", "Attendeance", "Assignment_Score", "Study_Hours_week"]]
y=df["Final_Marks"]
```

```python
x
```

```python
y
```

```python
model = LinearRegression()
model.fit(x, y)
print("Intercept:", model.intercept_)
```

```python
coefficient_table=pd.DataFrame({'Feature': x.columns, 'Coefficient': model.coef_})
```

```python
coefficient_table
```

```python
new_student=pd.DataFrame({
    "Study Hours_day":[6],
    "Attendeance":[82],
    "Assignment_Score":[72],
    "Study_Hours_week":[42]
})
 
prediction = model.predict(new_student)
print("Predicted Final Marks for the new student:", prediction[0])
```
